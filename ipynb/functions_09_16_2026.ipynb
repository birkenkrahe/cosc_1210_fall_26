## Parameters and arguments
### Functions with one parameter

````python
# function definition
def greet(name):  # `name` is a parameter
    print(f"Hello, {name}.")   # prints Hello, [name]
    print("Hello, " + name + " , again.")  

# function calls
greet("Jim")  # "Jim" is an argument
greet("Joe")
````

As for names, no Python objects (variables, functions, lists, ...) can use reserved names or violate any of the other rules.

````python
def hypotenuse(a,b):
    """Print the hypotenuse of a right triangle.

    a, b -- the lengths of the two legs (numbers)
    Prints the hypotenuse; returns nothing.
    """
    c = (a**2 + b**2) ** 0.5
    print(c)

hypotenuse(3,4)
````

````python
def discount(price,percent):
    """Print the sale price after taking a percentage off.

    price   -- the original price (a number)
    percent -- the discount as a whole number, so 20 means 20% off
    Prints the reduced price; returns nothing.
    """
    print(price - price * percent / 100)

discount(50,20)
````

## Generalizing the `rule()` function

````python
def rule(length=24):
    print('-' * length) 
````

````python
rule()
````

````python
def notice(text=""): # default: no notice
    rule(len(text))  # pass the length of the text to rule
    print(text)
    rule(len(text))
````

````python
notice("Classed cancelled on Friday and Monday!")
````

Process:

main() -> notice() -> rule() -> print() -> rule() -> main()

## Practice: `receipt` function

Write two functions:
1. `rule()` print a line of 24 dashes
2. `receipt(item, price, qty) - prints, in order:
    - a rule
    - the line `[qty x item]`
    - the line `Total: $[price x qty]
    - a rule
    - Use f-string for the two output lines
3. Provide docstrings for both functions
4. Call `receipt("coffee",3.5,2)`

Sample output:
```
------------------------
2 x coffee
Total: $7.0
------------------------
```


````python
def rule():
    """ Print a line of 24 dashes.
    [this is a multiline comment]
    """
    print("-" * 24)

def receipt(item, price, qty):
    """Print a receipt for one item.

    item  -- a sold item (a string)
    price -- the item price (a number)
    qty   -- the number of items sold (a number)
    Prints the <qty> x <item> line and the Total line (price * qty);
    returns nothing.
    Calls rule() for the top and bottom lines.
    """
    rule()
    print(f"{qty} x {item}\nTotal: ${price * qty}")
    rule()

receipt("coffee", 3.5, 2)
````

````python
help(receipt)
````

## Repetition with loops: `for`, `range()`

````python
for i in range(3): # range produces numbers
    print(i)
````

The `range` function cannot be called, it can only be looped over.

````python
range(3)
````

````python
help(range())
````

````python
del(range)
range(3)
````

````python
# list comprehension
[print(i, end=' ') for i in range(3)];
````

````python
for n in range(5):
    print(n+1, "squared is", (n+1)**2)
````

````python
for _ in range(3):      # _ is a dummy loop variable (not used)
    print("welcome!")
````

````python
# the last loop as a list comprehension
[print("welcome") for _ in range(3)];
````

````python
print(type([print("welcome") for _ in range(3)]))
````
