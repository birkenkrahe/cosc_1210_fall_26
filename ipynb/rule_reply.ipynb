## 2. A small engine

`respond` tries the rules in order: the first one that matches
answers. If no rule matches, the fallback answers. The comments (A),
(B) and (C) mark the three places where an extension can go.

```python
def respond(sentence):
    """Return the reply of the first rule that matches, else a fallback."""
    # (A) before the rules
    reply = rule_reply("^i am (.*)", "Why are you ", sentence)
    if reply is None:
        reply = rule_reply("i need (.*)", "Why do you need ", sentence)
    if reply is None:
        # (B) no rule matched
        reply = "Please go on."
    # (C) after the reply
    return reply
```

```python
print(respond("I'm tired."))
print(respond("Well, I need coffee!"))
print(respond("Hello."))
print(respond("I AM TIRED!"))
```
