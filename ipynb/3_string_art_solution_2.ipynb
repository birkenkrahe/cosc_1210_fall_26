## String Art



**Problem:** Build the poster below in Python. Respect the following
rules:

1. Top and bottom rows are one character repeated with `*`
2. Each line with text has to be indented by repeating single empty
   spaces

```
==============================
          BAKE  SALE
       Friday  at  3  pm
          Room  204
==============================
      12 dozen  for  $5
        SEE YOU THERE
```

## Solution



### What length does the poster have?



````python
print(len("=============================="))
````

### The top and bottom line of the poster



````python
print("=" * 30)
````

### Counting the leading spaces



Find out how many empty spaces there are before the first letter in
each line of text:



````python
print("          BAKE  SALE".find('B'))
print("       Friday  at  3  pm".find('F'))
print("          Room  204".find('R'))
print("      12 dozen  for  $5".find('1'))
print("        SEE YOU THERE".find('S'))
````

But to do this, you have to know what the first letter is that you're
looking for. Better to use a statement that tells you when the empty
spaces end.

Define a list that holds the text lines:



````python
text = [
    "          BAKE  SALE",
    "       Friday  at  3  pm",
    "          Room  204",
    "      12 dozen  for  $5",
    "        SEE YOU THERE"
]
````

Go through every line, count characters and return the number of
characters that are empty (before the first word):



````python
for line in text:
    count = 0
    for character in line:
        count+=1
        if not character == ' ':
            print(count-1)
            break
````

### Part 1 - Build the poster



The last two text lines are not needed yet. Put the printing into a
function so that the code does not have to be repeated:



````python
def poster():
    print("=" * 30)
    print(" " * 10 + "Bake   Sale")
    print(" " * 7  + "Friday  at  3  pm")
    print(" " * 10 + "Room  204")
    print("=" * 30)

poster()
````

### Part 2 - Add a count line



`"12  dozen  for  $5"` must not be one string but must be composed by
numbers and text using the `str` function:



````python
poster()
print(" " * 6 + str(12) + " dozen  for  $" + str(5))
````

### Part 3 - Shout it



Add `SEE YOU THERE` starting from the lowercase string and turning it
uppercase in your code:



````python
poster()
print(" " * 6 + str(12) + " dozen  for  $" + str(5))
print(" " * 8 + "see you there".upper())
````

### An alternative method: `str.center`



````python
help(str.center)
````

````python
def poster2():
    print("=" * 30)
    print("Bake   Sale".center(30,' '))
    print("Friday  at  3  pm".center(30,' '))
    print("Room  204".center(30,' '))
    print("=" * 30)
    print((str(12) + " dozen  for  $" + str(5)).center(30,' '))
    print("see you there".upper().center(30,' '))

poster2()
````
