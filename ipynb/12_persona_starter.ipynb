## Setup: code from the lists lesson

Run this cell first. `respond` needs these two functions.

```python
import re

def normalize(sentence):
    """Return sentence in lowercase, with "i'm"/"im" written out, without end punctuation."""
    sentence = sentence.lower()
    sentence = re.sub(r"\bi'?m\b", "i am", sentence)
    return sentence.strip(".!?")

def next_reply(replies):
    """Return the first reply and move it to the end of the list."""
    reply = replies.pop(0)
    replies.append(reply)
    return reply
```

## respond takes the script (demo)

Demo: run the cell and read the code. You need `reflections` and `reflect` from above.

```python
def respond(sentence, script):
    """Return the reply of the first rule whose pattern matches sentence, else a fallback reply."""
    sentence = normalize(sentence)
    for rule in script["rules"]:
        match = re.search(rule["pattern"], sentence)
        if match is not None:
            reply = next_reply(rule["replies"])
            if "{x}" in reply:
                rest = reflect(match.group(1), script["reflections"])
                reply = reply.replace("{x}", rest)
            return reply
    return next_reply(script["fallback"])

print(respond("I need my mother.", script))
print(respond("I'm awake.", script))
print(respond("Nice weather.", script))
```
